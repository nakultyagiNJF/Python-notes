# `Pyplot`

> * Most of the Matplotlib utilities lies under the `pyplot` submodule, and are usually imported under the `plt` alies

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
xpoints = np.array([0, 6])
ypoints = np.array([0, 250])

plt.plot(xpoints, ypoints)
plt.show()
```

# Plotting x and y points

```python
xpoints = np.array([1, 8])
ypoints = np.array([5, 13])

plt.plot(xpoints, ypoints, color='green', linestyle='dashed', linewidth=3)
plt.show()
```

```python
plt.plot(xpoints, ypoints, 'o')
plt.show()
```

Multiple Points

```python
xpoints = np.array([1, 2, 6, 8, 10])
ypoints = np.array([3, 8, 1, 10, 5])

plt.plot(xpoints, ypoints, color='red', linestyle='dashed', linewidth=2)
plt.show()
```

Defalut X-points

```python
ypoints = np.array([10, 23, 4, 2, 7, 8])

plt.plot(ypoints)
plt.show()
```

# Markers

```python
ypoints = np.array([5, 2, 1, 9, 6, 12])

plt.plot(ypoints, marker='>')
plt.show()
```

```python
"""
*marker     description
.            point
,            pixel
o            circle
v            triangle_down
^            triangle_up
<            triangle_left
>            triangle_right
1            tri_down
2            tri_up
3            tri_left
4            tri_right
s            square
p            pentagon
*            star
H            hexagon1
h            hexagon2
D            diamond
d            thin_diamond
X            x (filled)
x            X
+            Plus
p            plus (filled)
|            vertical line
-            horizontal line
"""
```

# Format String `fmt`

> fmt = marker | line | color

```python
ypoints = np.array([5, 2, 1, 9, 6, 12])

plt.plot(ypoints, 'o-.m')
plt.show()
```

```python
"""
*line syntax    description
-               solid line
:               dotted line
--              dashed line
-.              dash-dot line
"""
```

```python
"""
*color syntax    description
'b'              blue
'g'              green
'r'              red
'c'              cyan
'm'              magenta
'y'              yellow
'k'              black
'w'              white
"""
```

# Marker Size

```python
ypoints = np.array([5, 2, 1, 9, 6, 12])

plt.plot(ypoints, 'o-.m', ms=10)
plt.show()
```

# Marker Color

```python
ypoints = np.array([5, 2, 1, 9, 6, 12])

plt.plot(ypoints, marker='o', ms=10, mec='r')
plt.show()
```

# Marker Face color

```python
ypoints = np.array([5, 2, 1, 9, 6, 12])

plt.plot(ypoints, marker='o', ms=10, mec='g', mfc='r')
plt.show()
```

# Using Hex Color

```python
ypoints = np.array([5, 2, 1, 9, 6, 12])

plt.plot(ypoints, marker='o', ms=10, mec='#2c3e50', mfc='#2c3e50')
plt.show()
```

# Using Full Color Name

```python
ypoints = np.array([5, 2, 1, 9, 6, 12])

plt.plot(ypoints, marker='o', ms=10, mec='black', mfc='HotPink')
plt.show()
```
